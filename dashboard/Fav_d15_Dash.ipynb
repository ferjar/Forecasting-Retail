# Archivos para dashboard



```python
#Carga de datos
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd

cols = ['time_idx', 'date', 'store_nbr', 'item_nbr', 'city', 'state',
        'type', 'cluster', 'family', 'class', 'perishable']
panel = pd.read_parquet('/content/drive/MyDrive/TFM/favorita_panel_corregido.parquet', columns=cols)
panel = panel.rename(columns={'type': 'store_type'})
print(panel.shape)
```

```python
#Crear archivo de fechas
fechas = panel[['time_idx', 'date']].drop_duplicates().sort_values('time_idx').reset_index(drop=True)
fechas.to_csv('/content/drive/MyDrive/TFM/time_idx_fechas.csv', index=False)
print('fechas:', fechas.shape)
fechas.head()
```

```python
#Crear archivo de metadatos
metadata = panel[['store_nbr', 'item_nbr', 'city', 'state', 'store_type',
                   'cluster', 'family', 'class', 'perishable']].drop_duplicates().reset_index(drop=True)
metadata.to_csv('/content/drive/MyDrive/TFM/metadata_series.csv', index=False)
print('metadata_series:', metadata.shape, '(esperado cercano a 3.981 series retenidas)')
metadata.head()
```
